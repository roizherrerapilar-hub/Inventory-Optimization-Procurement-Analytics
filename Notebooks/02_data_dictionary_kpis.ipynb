# 02 - Data Dictionary & KPI Definition

The purpose of this notebook is to document the available variables, define the key business metrics and establish the business questions that will guide the analysis throughout the project.


## Business Context

The objective of this project is to improve inventory management and support procurement decisions by increasing visibility across the supply chain.

To achieve this, the analysis focuses on inventory levels, demand, replenishment activities and supplier performance with the following objectives:

- Improve inventory efficiency.
- Reduce inventory holding costs.
- Optimise procurement decisions.
- Increase supply chain visibility.
- Explore the potential impact of exchange rate fluctuations.

## Business Questions

### Inventory

- Which products maintain the highest inventory levels?
- Are there products with excess inventory?
- Which products require replenishment more frequently?
- How does demand compare with the available inventory?

### Procurement

- Which suppliers generate the highest procurement volume?
- Which suppliers have the longest lead times?
- Which suppliers account for the highest procurement spend?

### Warehouses

- Which warehouses manage the highest inventory levels?
- Which warehouses record the highest sales volume?

### Forecast

- What is the difference between forecast demand and actual demand?
- How accurate are demand forecasts?

### External Factors

- How could exchange rate fluctuations affect procurement costs?

## Inventory KPIs

### Inventory Value

Represents the total financial value of the available inventory.

**Formula**

Inventory Level × Unit Cost

---

### Inventory Coverage

Measures how many days the available inventory can support forecast demand.

**Formula**

Inventory Level ÷ Demand Forecast

---

### Replenishment Frequency

Measures how frequently inventory is replenished.

**Formula**

Number of records where Order Quantity > 0

## Procurement KPIs

### Procurement Spend

Represents the total purchasing cost generated through replenishment activities.

**Formula**

Order Quantity × Unit Cost

---

### Average Supplier Lead Time

Measures the average delivery time required by each supplier.

**Formula**

Average of Supplier Lead Time Days

---

### Spend by Supplier

Shows how procurement spend is distributed across suppliers, helping identify strategic suppliers and purchasing concentration.

## Forecast KPIs

### Forecast Error

Measures the difference between actual demand and forecast demand.

**Formula**

Units Sold − Demand Forecast

---

### Forecast Accuracy

Measures how closely forecast demand matches actual demand, providing an indicator of demand planning performance.

## Exchange Rate API Integration

To extend the analytical capabilities of the project, an exchange rate API was integrated.

Although it is not included in the final Power BI dashboard, this functionality enables future financial analyses such as:

- Evaluating the impact of USD exchange rate fluctuations on procurement costs.
- Comparing purchasing costs across different currencies.
- Assessing inventory exposure to currency risk.
